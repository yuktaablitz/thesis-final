# 01 — Data Audit

Run this first after infrastructure is set up. Purpose: sanity-check every
raw source before building the master table.

- Download/refresh each source via `src.data.*`
- Plot coverage (row count over time) per source — find gaps
- Confirm timestamp dtypes are UTC-aware everywhere
- Spot-check a few known events (e.g. a halving, a major hack) land where expected
